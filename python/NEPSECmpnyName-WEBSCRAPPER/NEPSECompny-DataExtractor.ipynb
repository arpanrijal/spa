```python
import pandas as pd
from bs4 import BeautifulSoup
```

```python
d = {"S.N.":[],	"Name of Company":[],	"Symbol":[],	"ISIN":[],	"ISIN Type":[]}
file = "NEPSECompanyExtractor" + ".html"
filepath = f"../DATA-HTML-STOCK/webScrapped-htmlfiles/NEPSECompanyName/{file}"
try:
    with open(f"{filepath}",  encoding="utf-8") as f:
        html_doc = f.read()
    
    soup = BeautifulSoup(html_doc, 'html.parser')    
    rows = soup.find('tbody').find_all('tr')
    for row in rows:
        cols = row.find_all('td')
        
        if len(cols) >= 4:
            d['S.N.'].append(cols[0].get_text().strip())
            d['Name of Company'].append(cols[1].get_text().strip())
            d['Symbol'].append(cols[2].get_text().strip())
            d['ISIN'].append(cols[3].get_text().strip())
            d['ISIN Type'].append(cols[4].get_text().strip())
    print(f"{file}: {len(rows)} rows")
except Exception as e:
    print(f"{e}")
csv_filename = file.replace('.html', '.csv')
csv_file_path = f"../DATA-HTML-STOCK/NEPSECompany/{csv_filename}"
df = pd.DataFrame(data=d)
df.to_csv(f"{csv_file_path}", index=False)
print("done")
```
