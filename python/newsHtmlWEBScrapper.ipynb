```python
from selenium import webdriver
from selenium.webdriver.common.by import By
import pandas as pd
import time
```

```python
driver = webdriver.Edge()
driver.set_page_load_timeout(6)
query = "sypnl"
newsname="sypnlnewshtml"
try:
    driver.get(f'https://www.sharesansar.com/company/{query}')
except:
    pass
time.sleep(1)
element = driver.find_elements(By.LINK_TEXT, "News")
element[1].click()
time.sleep(3)
all_rows=[]
page=1
while True:
    print(f"Page {page}...", end=" ")
    rows = driver.find_elements(By.CSS_SELECTOR, "#myTableCNews tbody tr")
    for row in rows:
        all_rows.append(row.get_attribute('outerHTML'))
    print(f"✓ {len(rows)} rows")
    next_button = driver.find_element(By.ID, "myTableCNews_next")
    if "disabled" in next_button.get_attribute("class"):
        break
    next_button.click()
    time.sleep(2)
    page += 1
html_content=f'''<table id="myTableCPriceHistory">
<tbody>
{''.join(all_rows)}
</tbody>
</table>
'''
with open(f'STOCKNEPSE/webScraphtmlfiles/{newsname}.html', 'w', encoding='utf-8') as f:
    f.write(html_content)
print(f"\n✓ Done! {len(all_rows)} rows saved to HTML")
driver.close()
```
