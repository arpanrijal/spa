```python
import time
from selenium import webdriver
from selenium.webdriver.common.by import By
```

```python
driver = webdriver.Edge()
driver.set_page_load_timeout(3)

query="NEPSECompanyExtractor"

try:
    driver.get(f"https://www.sharesansar.com/cdsc")
except:
    pass

all_rows=[]
page=1

while True:
    print(f"Page {page}...", end=" ")
    rows = driver.find_elements(By.CSS_SELECTOR, "#myTable tbody tr")
    for row in rows:
        all_rows.append(row.get_attribute('outerHTML'))
    nxt_btn = driver.find_element(By.ID, "myTable_next")
    if "disabled" in nxt_btn.get_attribute("class"):
        break
    nxt_btn.click()
    time.sleep(2)
    page += 1

html_content = f'''<table id="myTableCPriceHistory">
<tbody>
{''.join(all_rows)} # Output: <tr>row1</tr><tr>row2</tr><tr>row3</tr>
</tbody>
</table>'''

with open(f'STOCKNEPSE/webScraphtmlfiles/{query}.html', 'w', encoding='utf-8') as f:
    f.write(html_content)
    
print(f"\n✓ Done! {len(all_rows)} rows saved to HTML")
driver.close()
```
