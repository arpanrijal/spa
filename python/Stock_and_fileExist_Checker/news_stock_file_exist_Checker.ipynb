```python
import pandas as pd
import os
```

```python
filepath = f"../DATA-HTML-STOCK/NEPSECompany/NEPSECompanyExtractor.csv"
df = pd.read_csv(filepath)
symbols = df['Symbol']
stock_no_html = []
stock_no_csv = []

for symbol in symbols:
    html_path = f"../DATA-HTML-STOCK/webScrapped-htmlfiles/news-WEB-SCRAP-htmlfile/{symbol}news.html" 
    csv_path = f"../DATA-HTML-STOCK/NEPSENEWS/{symbol}news.csv"
    if not os.path.exists(html_path):
        
        print(f"{symbol} html file not exist")
        stock_no_html.append(symbol)
        
        if not os.path.exists(csv_path):
            print(f"{symbol} csv file not exist")
            stock_no_csv.append(symbol)
        else:
            print(f"{symbol} csv file exist")
    

print(f"stock_no_html: {stock_no_html}")
print(f"stock_no_csv: {stock_no_csv}")
```
