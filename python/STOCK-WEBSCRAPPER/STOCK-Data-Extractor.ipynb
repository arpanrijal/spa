```python
from bs4 import BeautifulSoup
import pandas as pd
import os
```

```python
d = {'S.N.': [], 'Date': [], 'Open': [], 'High': [], 'Low': [], 'Close': [], '% Change': [], 'Qty': [], 'Turnover': []}
query = "NEPSECompanyExtractor"
companyDetails = pd.read_csv(f"../DATA-HTML-STOCK/NEPSECompany/{query}.csv")
symbols = companyDetails['Symbol']
for symbol in symbols:
    
    filepath = f"../DATA-HTML-STOCK/webScrapped-htmlfiles/NEPSE-StockWebScrap/{symbol}.html"
    if not os.path.exists(filepath):
        print(f"{symbol} not exist")
        continue
    
    try:
        with open(f"{filepath}", encoding="utf-8") as f:
            html_doc = f.read()
        
        soup = BeautifulSoup(html_doc, 'html.parser')        
        rows = soup.find('tbody').find_all('tr')
        
        for row in rows:
            cols = row.find_all('td')
    
            if len(cols) >= 9:
                d['S.N.'].append(cols[0].get_text().strip())
                d['Date'].append(cols[1].get_text().strip())
                d['Open'].append(cols[2].get_text().strip())
                d['High'].append(cols[3].get_text().strip())
                d['Low'].append(cols[4].get_text().strip())
                d['Close'].append(cols[5].get_text().strip())
                d['% Change'].append(cols[6].get_text().strip())
                d['Qty'].append(cols[7].get_text().strip())
                d['Turnover'].append(cols[8].get_text().strip())
        
        print(f"{symbol}: {len(rows)} rows")
        
    except Exception as e:
        print(f"{e}")
    csv_filename = f"{symbol}.csv"
    savepath = f"../DATA-HTML-STOCK/NEPSEDATA/{csv_filename}"
    df = pd.DataFrame(data=d)
    df.to_csv(f"{savepath}", index=False)
    print(f"✓ {symbol} completed \n")
print("All Done")
```
