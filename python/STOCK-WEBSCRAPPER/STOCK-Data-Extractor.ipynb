```python
from bs4 import BeautifulSoup
import pandas as pd
import os
```

```python
query = "NEPSECompanyExtractor"
companyDetails = pd.read_csv(f"../DATA-HTML-STOCK/NEPSECompany/{query}.csv")
symbols = companyDetails['Symbol']

for symbol in symbols:

    print(f"\nExtracting {symbol}...")

    filepath = f"../DATA-HTML-STOCK/webScrapped-htmlfiles/NEPSE-StockWebScrap/{symbol}.html"

    if not os.path.exists(filepath):
        print(f"{symbol} not exist")
        continue

    try:
        with open(filepath, encoding="utf-8") as f:
            html_doc = f.read()

        soup = BeautifulSoup(html_doc, 'html.parser')

        table = soup.find('table', {'id': 'myTableCPriceHistory'})
        if table is None:
            print(f"{symbol}: table not found")
            continue

        rows = table.find('tbody').find_all('tr')

        data = []

        for row in rows:
            cols = row.find_all('td')

            if len(cols) >= 9:
                data.append({
                    'Date': cols[1].get_text(strip=True),
                    'Open': cols[2].get_text(strip=True),
                    'High': cols[3].get_text(strip=True),
                    'Low': cols[4].get_text(strip=True),
                    'Close': cols[5].get_text(strip=True),
                    '% Change': cols[6].get_text(strip=True),
                    'Qty': cols[7].get_text(strip=True),
                    'Turnover': cols[8].get_text(strip=True)
                })

        df = pd.DataFrame(data)

        df['Date'] = pd.to_datetime(df['Date'], errors='coerce')

        numeric_cols = ['Open', 'High', 'Low', 'Close', '% Change', 'Qty', 'Turnover']

        for col in numeric_cols:
            df[col] = pd.to_numeric(
                df[col].astype(str).str.replace(',', ''),
                errors='coerce'
            )

        df = df.dropna()

        df = df.sort_values(by='Date')
        df = df.drop_duplicates(subset=['Date'], keep='last')

        df.reset_index(drop=True, inplace=True)
        df.insert(0, 'S.N.', df.index + 1)

        print(f"{symbol}: cleaned {len(df)} rows")

        savepath = f"../DATA-HTML-STOCK/NEPSEDATA/{symbol}.csv"
        df.to_csv(savepath, index=False)

        print(f"✓ {symbol} completed\n")

    except Exception as e:
        print(f"{symbol} error: {e}")

print("All Done")
```
