```python
from selenium import webdriver
from selenium.webdriver.common.by import By
import pandas as pd
import time
```

```python
df = pd.read_csv("STOCKNEPSE/NEPSECompany/NEPSECompanyExtractor.csv")

symbols = df['Symbol'].tolist()

print(f"Found {len(symbols)} companies to scrape")

driver = webdriver.Edge()
driver.set_page_load_timeout(6)

query = "crest"

try:
    driver.get(f"https://www.sharesansar.com/company/{query}")
except:
    pass

time.sleep(1)

driver.find_element(By.LINK_TEXT, "Price History").click()
time.sleep(3)

all_rows = []
page = 1

while True:
    print(f"Page {page}...", end=" ")
    
    rows = driver.find_elements(By.CSS_SELECTOR, "#myTableCPriceHistory tbody tr")
    
    for row in rows:
        all_rows.append(row.get_attribute('outerHTML'))
    # ["<tr>row1</tr>", "<tr>row2</tr>", "<tr>row3</tr>"]
    print(f"✓ {len(rows)} rows")
    
    next_button = driver.find_element(By.ID, "myTableCPriceHistory_next")
    if "disabled" in next_button.get_attribute("class"):
        break
    
    next_button.click()
    time.sleep(2)
    page += 1

html_content = f'''<table id="myTableCPriceHistory">
<tbody>
{''.join(all_rows)} # Output: <tr>row1</tr><tr>row2</tr><tr>row3</tr>
</tbody>
</table>'''

with open(f'STOCKNEPSE/webScraphtmlfiles/{query}.html', 'w', encoding='utf-8') as f:
    f.write(html_content)

print(f"\n✓ Done! {len(all_rows)} rows saved to HTML")
driver.close()
```

```python

# Load the CSV with company symbols
df = pd.read_csv("STOCKNEPSE/NEPSECompany/NEPSECompanyExtractor.csv")

# Get list of symbols
symbols = df['Symbol'].tolist()

print(f"Found {len(symbols)} companies to scrape")

driver = webdriver.Edge()
driver.set_page_load_timeout(6)

for query in symbols:
    print(f"\n{'='*50}")
    print(f"Scraping: {query}")
    print(f"{'='*50}")
    
    try:
        driver.get(f"https://www.sharesansar.com/company/{query}")
    except:
        pass
    
    time.sleep(1)
    
    try:
        driver.find_element(By.LINK_TEXT, "Price History").click()
        time.sleep(3)
        
        all_rows = []
        page = 1
        
        while True:
            print(f"Page {page}...", end=" ")
            
            rows = driver.find_elements(By.CSS_SELECTOR, "#myTableCPriceHistory tbody tr")
            
            for row in rows:
                all_rows.append(row.get_attribute('outerHTML'))
            
            print(f"✓ {len(rows)} rows")
            
            next_button = driver.find_element(By.ID, "myTableCPriceHistory_next")
            if "disabled" in next_button.get_attribute("class"):
                break
            
            next_button.click()
            time.sleep(2)
            page += 1
        
        html_content = f'''<table id="myTableCPriceHistory">
<tbody>
{''.join(all_rows)}
</tbody>
</table>'''
        
        with open(f'STOCKNEPSE/webScraphtmlfiles/{query}.html', 'w', encoding='utf-8') as f:
            f.write(html_content)
        
        print(f"✓ {query}: {len(all_rows)} total rows saved")
        
    except Exception as e:
        print(f"✗ Error with {query}: {e}")
        continue

driver.close()
print("\n✓ All companies scraped!")
```
