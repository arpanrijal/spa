```python
import os
import pandas as pd

filepath = "../DATA-HTML-STOCK/NEPSECompany/NEPSECompanyExtractor.csv"
df = pd.read_csv(filepath)
symbols = df['Symbol'].tolist()

stock_ignored = ["CORBLP", "GFCLPO", "HBLPO", "LFCPO", "SBIPO", "GABLPO", "AMFIPO", "NMLBS", "ILFCPO"]
stock_no_news = ["HATHPO", "BUDBLP", "CFCLPO", "CORBLP", "EBLPO", "HAMROP", "HGIPO", "JFLPO", "JBNLPO", "KADBLP", "KNBLPO", "CEFLPO", "LBLPO", "MFILPO", "MIDBLP", "NBBLPO", "NABBPO", "NBBPO", "PFLPO", "PRINPO", "PURBLP", "SBBLJP", "SIFCPO", "SILPO", "SMFDBP", "SYFLPO", "TNBLPO", "TDBLPO", "UFLPO", "WDBLPO", "WMBFPO", "HLBSLP"]
stock_not_qualify = ['ALICLP', 'BOKLPO', 'BFCPO', 'BHBLPO', 'BNL', 'CCBLPO', 'CZBILP', 'CBLPO', 'DDBLPO', 'DBBLPO', 'EDBLPO', 'FMDBLP', 'GDBLPO', 'GBBLPO', 'GABL', 'GBIMEP', 'GMFILP', 'GLICLP', 'GUFLPO', 'ICFCPO', 'IGIPO', 'JEFLPO', 'JBBLPO', 'KBBLPO', 'KMCDBP', 'KSBBLP', 'KRBLPO', 'CEFL', 'KBLPO', 'LFC', 'MBLPO', 'MLBLPO', 'MEGAPO', 'MIDBL', 'MDBPO', 'MMDBL', 'MNBBLP', 'MPFLPO', 'NBSL', 'NLICLP', 'NILPO', 'NABB', 'NCDBPO', 'NCCBPO', 'GBLBSP', 'NLICP', 'NICAP', 'NIDC', 'NMBPO', 'ODBL', 'ODBLPO', 'PRVUPO', 'PCBLP', 'PLICPO', 'PROFLP', 'PICLPO', 'RLFLPO', 'RMDCPO', 'SICPO', 'SHBL', 'SNMAPO', 'SADBLP', 'SICLPO', 'SBLPO', 'SINDUP', 'SRS', 'SFFILP', 'SRBLPO', 'SLICLP', 'SWBBLP', 'SLBBLP', 'UICPO', 'AMFI', 'CBBLPO', 'FOWADP', 'GILBPO', 'GRDBLP', 'JSLBBP', 'WOMIPO']
#79 ota stock is not qualifed/have less than 500 rows. 174 stock matrai xa which we can actuall work with currently.


nofile_skip = []
stock_not_qualify = []
stock_qualify = []

for symbol in symbols:

    if symbol in stock_ignored:
        continue

    if symbol in stock_no_news:
        continue

    file_path = f"../DATA-HTML-STOCK/FinalDataSet/{symbol}.csv"

    if not os.path.exists(file_path):
        nofile_skip.append(symbol)
        continue

    file_df = pd.read_csv(file_path)
    filerowsize = len(file_df)

    if filerowsize < 500:
        stock_not_qualify.append(symbol)
    else:
        stock_qualify.append(symbol)

print(f"Not qualified : {stock_not_qualify}, {len(stock_not_qualify)}")
print(f"No File skip : {nofile_skip},  {len(nofile_skip)}")
```
